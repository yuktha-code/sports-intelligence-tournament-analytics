```python
# ==========================================
# ðŸŒ¡ï¸ CELL 3 â€” Venue, Weather & Pitch Matrix
# ==========================================
# Based on July 2026 historical climate data for each host city
VENUES = {
    'New York/New Jersey': (0.30, 0.00, 1.00),   # Final venue â€” MetLife
    'Los Angeles':         (0.35, 0.05, 1.00),
    'Dallas':              (0.60, 0.15, 1.00),
    'Miami':               (0.70, 0.00, 0.95),
    'Atlanta':             (0.55, 0.07, 1.00),
    'Houston':             (0.75, 0.00, 0.95),
    'Kansas City':         (0.50, 0.23, 1.00),
    'Seattle':             (0.20, 0.00, 1.00),
    'San Francisco':       (0.20, 0.00, 1.00),
    'Philadelphia':        (0.35, 0.00, 1.00),
    'Boston':              (0.30, 0.00, 1.00),
    'Toronto':             (0.35, 0.00, 1.00),
    'Vancouver':           (0.20, 0.00, 1.00),
    'Mexico City':         (0.40, 2.24, 0.95),  # 2,240 m altitude
    'Guadalajara':         (0.45, 1.57, 1.00),
    'Monterrey':           (0.65, 0.54, 1.00),
}

GROUP_VENUES = {
    'A': 'Mexico City', 'B': 'Toronto', 'C': 'Los Angeles', 'D': 'Kansas City',
    'E': 'Dallas', 'F': 'Seattle', 'G': 'Atlanta', 'H': 'Houston',
    'I': 'Miami', 'J': 'San Francisco', 'K': 'Philadelphia', 'L': 'Boston',
}

KNOCKOUT_VENUE = 'New York/New Jersey'

def venue_modifier(team1, team2, stage='group'):
    grp = df.loc[team1, 'group'] if stage == 'group' else None
    city = GROUP_VENUES.get(grp, KNOCKOUT_VENUE) if stage == 'group' else KNOCKOUT_VENUE
    heat, alt, pitch = VENUES[city]

    ALTITUDE_ADAPTED = {'Mexico', 'Colombia', 'Ecuador', 'Bolivia', 'Peru', 'Argentina',
                        'Chile', 'Uruguay', 'Paraguay', 'Venezuela'}

    def _mod(team):
        t = df.loc[team]
        heat_pen  = heat * (t['coach_attack'] / 10) * 0.05   # High press suffers in heat
        alt_pen   = 0.0 if team in ALTITUDE_ADAPTED else alt * 0.015
        pitch_pen = (1 - pitch) * (t['avg_possession'] / 100) * 0.08
        return 1.0 - heat_pen - alt_pen - pitch_pen

    return _mod(team1), _mod(team2)

print('âœ… Venue, weather, and pitch condition matrix loaded.')
```

```python
# ==========================================
# âš¡ CELL 4 â€” Core Strength Calculation Engine
# ==========================================
WEIGHTS = {
    'elo':            0.28,
    'form_pts':       0.18,
    'attack_index':   0.12,
    'defense_index':  0.12,
    'squad_depth':    0.10,
    'experience':     0.08,
    'coach_attack':   0.05,
    'coach_defense':  0.05,
    'h2h_advantage':  0.02,
}

def normalise_col(series):
    return (series - series.min()) / (series.max() - series.min() + 1e-9)

norm_df = df.copy()
for col in WEIGHTS:
    norm_df[col] = normalise_col(df[col])

def age_factor(age):
    if 25.5 <= age <= 28.0:
        return 1.00
    elif age > 28.0:
        return 1.00 - (age - 28.0) * 0.015
    else:
        return 1.00 - (25.5 - age) * 0.010

def expected_goals(team):
    t = df.loc[team]
    base_lambda = 1.25
    poss_mult   = t['avg_possession'] / 50.0
    atk_mult    = t['attack_index'] / 6.5
    age_mult    = age_factor(t['avg_age'])
    inj_pen     = 1.0 - (t['injury_risk'] * 0.3)
    return base_lambda * poss_mult * atk_mult * age_mult * inj_pen

def composite_strength(team):
    score = sum(WEIGHTS[col] * norm_df.loc[team, col] for col in WEIGHTS)
    score *= age_factor(df.loc[team, 'avg_age'])
    score *= (1.0 - df.loc[team, 'injury_risk'] * 0.15)
    return score

df['strength'] = df.index.map(composite_strength)
df['xG']       = df.index.map(expected_goals)

print('âœ… Composite strength scores computed.')
```

```python
# ==========================================
# ðŸŽ² CELL 5 â€” Match Simulation Engine
# ==========================================
def injury_event(team):
    return np.random.random() < df.loc[team, 'injury_risk']

def simulate_match(team1, team2, stage='group', neutral=False):
    t1 = df.loc[team1]
    t2 = df.loc[team2]

    vm1, vm2 = venue_modifier(team1, team2, stage)
    inj1 = 0.85 if injury_event(team1) else 1.0
    inj2 = 0.85 if injury_event(team2) else 1.0

    foul_vol = (t1['foul_rate'] + t2['foul_rate']) / 28.0
    foul_vol = np.clip(foul_vol, 0.8, 1.3)

    str1 = df.loc[team1, 'strength'] * vm1 * inj1
    str2 = df.loc[team2, 'strength'] * vm2 * inj2

    xg1_raw = t1['xG'] * (t1['attack_index'] / (t2['defense_index'] + 0.5)) * foul_vol
    xg2_raw = t2['xG'] * (t2['attack_index'] / (t1['defense_index'] + 0.5)) * foul_vol

    ratio = str1 / (str1 + str2 + 1e-9)
    xg1 = np.clip(xg1_raw * ratio * 2.0, 0.3, 5.0) * inj1
    xg2 = np.clip(xg2_raw * (1 - ratio) * 2.0, 0.3, 5.0) * inj2

    h2h_adj = t1['h2h_advantage'] * 0.05
    xg1 += h2h_adj
    xg2 -= h2h_adj

    g1 = np.random.poisson(max(xg1, 0.1))
    g2 = np.random.poisson(max(xg2, 0.1))

    if g1 > g2:
        winner = team1
    elif g2 > g1:
        winner = team2
    else:
        if stage == 'group':
            winner = None
        else:
            form1 = df.loc[team1, 'form_pts'] / 30.0
            form2 = df.loc[team2, 'form_pts'] / 30.0
            pen_prob1 = 0.5 + 0.1 * (form1 - form2)
            pen_prob1 = np.clip(pen_prob1, 0.35, 0.65)
            winner = team1 if np.random.random() < pen_prob1 else team2

    return g1, g2, winner

print('âœ… Match engine ready. Testing individual simulation below:')
g1, g2, w = simulate_match('France', 'Argentina', stage='knockout')
print(f'   Test Match: France {g1}â€“{g2} Argentina â†’ Winner: {w}')
```

```python
# ==========================================
# ðŸ“Š CELL 6 â€” Group Stage Simulation & Standings
# ==========================================
GROUPS = {
    'A': ['Mexico', 'South Korea', 'South Africa', 'Czechia'],
    'B': ['Canada', 'Switzerland', 'Qatar', 'Bosnia-Herzegovina'],
    'C': ['Brazil', 'Morocco', 'Scotland', 'Haiti'],
    'D': ['USA', 'Australia', 'Paraguay', 'Turkiye'],
    'E': ['Germany', 'Ecuador', 'Ivory Coast', 'Curacao'],
    'F': ['Netherlands', 'Japan', 'Sweden', 'Tunisia'],
    'G': ['Belgium', 'Iran', 'Egypt', 'New Zealand'],
    'H': ['Spain', 'Uruguay', 'Saudi Arabia', 'Cape Verde'],
    'I': ['France', 'Senegal', 'Norway', 'Iraq'],
    'J': ['Argentina', 'Austria', 'Algeria', 'Jordan'],
    'K': ['Portugal', 'Colombia', 'Uzbekistan', 'DR Congo'],
    'L': ['England', 'Croatia', 'Panama', 'Ghana'],
}

def simulate_group(group_name, teams):
    stats = {t: {'pts': 0, 'gd': 0, 'gf': 0, 'ga': 0} for t in teams}

    for i, t1 in enumerate(teams):
        for t2 in teams[i+1:]:
            g1, g2, winner = simulate_match(t1, t2, stage='group')
            stats[t1]['gf'] += g1; stats[t1]['ga'] += g2
            stats[t2]['gf'] += g2; stats[t2]['ga'] += g1
            stats[t1]['gd'] += (g1 - g2)
            stats[t2]['gd'] += (g2 - g1)
            if winner == t1:
                stats[t1]['pts'] += 3
            elif winner == t2:
                stats[t2]['pts'] += 3
            else:
                stats[t1]['pts'] += 1
                stats[t2]['pts'] += 1

    standings = sorted(
        stats.items(),
        key=lambda x: (x[1]['pts'], x[1]['gd'], x[1]['gf'], np.random.random()),
        reverse=True
    )
    return standings

# Run one test pass for visualization
np.random.seed(42)
print('\n' + '=' * 65)
print('SINGLE-PASS GROUP STAGE SIMULATION BREAKDOWN (Seed=42)')
print('=' * 65)

group_qualifiers = {}
all_third = []

for g, teams in GROUPS.items():
    standings = simulate_group(g, teams)
    group_qualifiers[g] = [standings[0][0], standings[1][0]]
    all_third.append((standings[2][0], standings[2][1]['pts'], standings[2][1]['gd'], standings[2][1]['gf']))
    print(f'\nGroup {g}:')
    for rank, (team, s) in enumerate(standings, 1):
        qual = 'âœ…' if rank <= 2 else ' '
        print(f'  {qual} {rank}. {team:<22} Pts:{s["pts"]} GD:{s["gd"]:+d} GF:{s["gf"]}')

all_third_sorted = sorted(all_third, key=lambda x: (x[1], x[2], x[3]), reverse=True)
best_third = [x[0] for x in all_third_sorted[:8]]
print(f'\nðŸ“Œ Best 8 third-place teams qualifying: {best_third}')
```

```python
# =============================================================================
# ðŸ† PART 2: KNOCKOUT BRACKET STRUCTURE, MONTE CARLO LOOP & FINAL DATA COMPILATION
# =============================================================================

# ==========================================
# ðŸ—ºï¸ CELL 7 â€” Knockout Bracket Matchmaking Engine
# ==========================================
def structure_round_of_32(qualifiers, best_third):
    """
    Pairs Group Winners (1st) and Runners-up (2nd) alongside the 8 best 3rd-place teams
    according to the 48-team tournament layout.
    """
    r32_matches = []

    # Extract qualifiers from the dictionary
    A1, A2 = qualifiers['A'][0], qualifiers['A'][1]
    B1, B2 = qualifiers['B'][0], qualifiers['B'][1]
    C1, C2 = qualifiers['C'][0], qualifiers['C'][1]
    D1, D2 = qualifiers['D'][0], qualifiers['D'][1]
    E1, E2 = qualifiers['E'][0], qualifiers['E'][1]
    F1, F2 = qualifiers['F'][0], qualifiers['F'][1]
    G1, G2 = qualifiers['G'][0], qualifiers['G'][1]
    H1, H2 = qualifiers['H'][0], qualifiers['H'][1]
    I1, I2 = qualifiers['I'][0], qualifiers['I'][1]
    J1, J2 = qualifiers['J'][0], qualifiers['J'][1]
    K1, K2 = qualifiers['K'][0], qualifiers['K'][1]
    L1, L2 = qualifiers['L'][0], qualifiers['L'][1]

    # 16 Knockout fixtures mapping group pathways + wildcard 3rd-places dynamically
    r32_fixtures = [
        (A1, best_third[0] if len(best_third)>0 else 'Czechia'),
        (B1, C2), (C1, best_third[1] if len(best_third)>1 else 'Scotland'),
        (D1, E2), (E1, best_third[2] if len(best_third)>2 else 'Ivory Coast'),
        (F1, G2), (G1, best_third[3] if len(best_third)>3 else 'Egypt'),
        (H1, I2), (I1, best_third[4] if len(best_third)>4 else 'Norway'),
        (J1, K2), (K1, best_third[5] if len(best_third)>5 else 'Algeria'),
        (L1, A2), (B2, F2), (D2, H2), (J2, L2),
        (best_third[6] if len(best_third)>6 else 'Sweden', best_third[7] if len(best_third)>7 else 'Australia')
    ]

    winners = []
    for t1, t2 in r32_fixtures:
        _, _, w = simulate_match(t1, t2, stage='knockout')
        winners.append(w)
    return winners

def run_knockout_stage(r32_winners):
    """
    Simulates consecutive single-elimination tournament paths.
    """
    # --- ROUND OF 16 ---
    r16_winners = []
    for i in range(0, len(r32_winners), 2):
        if i+1 < len(r32_winners):
            _, _, w = simulate_match(r32_winners[i], r32_winners[i+1], stage='knockout')
            r16_winners.append(w)
        else:
            r16_winners.append(r32_winners[i])

    # --- QUARTER-FINALS ---
    qf_winners = []
    for i in range(0, len(r16_winners), 2):
        if i+1 < len(r16_winners):
            _, _, w = simulate_match(r16_winners[i], r16_winners[i+1], stage='knockout')
            qf_winners.append(w)
        else:
            qf_winners.append(r16_winners[i])

    # --- SEMI-FINALS ---
    sf_winners = []
    for i in range(0, len(qf_winners), 2):
        if i+1 < len(qf_winners):
            _, _, w = simulate_match(qf_winners[i], qf_winners[i+1], stage='knockout')
            sf_winners.append(w)
        else:
            sf_winners.append(qf_winners[i])

    # --- THE FINAL ---
    if len(sf_winners) >= 2:
        g1, g2, champion = simulate_match(sf_winners[0], sf_winners[1], stage='knockout')
        runner_up = sf_winners[1] if champion == sf_winners[0] else sf_winners[0]
    else:
        champion = sf_winners[0] if len(sf_winners) > 0 else 'Unknown'
        runner_up = 'Unknown'

    return r16_winners, qf_winners, sf_winners, champion, runner_up

print("âœ… Knockout stage bracket engine defined.")
```

```python
# ==========================================
# ðŸŽ² CELL 8 â€” 10,000 Iteration Monte Carlo Simulation Loop
# ==========================================
N_SIMULATIONS = 10000

# Frequency counters for tracking progression metrics
champion_counts = Counter()
runner_up_counts = Counter()
semi_final_counts = Counter()
quarter_final_counts = Counter()
round_of_16_counts = Counter()

print(f"ðŸš€ Simulating {N_SIMULATIONS:,} complete World Cup tournaments...")

for _ in tqdm(range(N_SIMULATIONS)):
    # 1. Simulate Group Stages
    qualifiers = {}
    third_places = []
    for g, teams in GROUPS.items():
        standings = simulate_group(g, teams)
        qualifiers[g] = [standings[0][0], standings[1][0]]
        third_places.append((standings[2][0], standings[2][1]['pts'], standings[2][1]['gd'], standings[2][1]['gf']))

    # Process and rank 3rd place wildcards
    third_places_sorted = sorted(third_places, key=lambda x: (x[1], x[2], x[3]), reverse=True)
    best_third = [x[0] for x in third_places_sorted[:8]]

    # 2. Run Knockout Matches
    r32_winners = structure_round_of_32(qualifiers, best_third)
    r16_w, qf_w, sf_w, champ, runner = run_knockout_stage(r32_winners)

    # 3. Aggregate Statistical Trends
    champion_counts[champ] += 1
    runner_up_counts[runner] += 1
    for t in sf_w: semi_final_counts[t] += 1
    for t in qf_w: quarter_final_counts[t] += 1
    for t in r16_w: round_of_16_counts[t] += 1

print("\nâœ… Simulations completed successfully.")
```
